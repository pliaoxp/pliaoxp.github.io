```python
import os
import re

root_folder = '/Users/yliao13/Documents/GitHub/pliaoxp.github.io'

old_domain = "https://econweb.rutgers.edu/yl1114"
new_script = '<script>\n   const linkhost = "https://pliaoxp.github.io";\n</script>\n'


def add_script_and_replace_links(filepath):
    with open(filepath, 'r', encoding='utf-8', errors='ignore') as f:
        content = f.read()

    # Check if script is already added, to avoid duplicates
    if new_script.strip() not in content:
        # Insert the script block after the <head> tag if present, else at beginning
        head_pos = content.lower().find('<head>')
        if head_pos != -1:
            insert_pos = head_pos + len('<head>')
            content = content[:insert_pos] + '\n' + new_script + content[insert_pos:]
        else:
            # If no <head>, add at beginning
            content = new_script + content

    # Regex to find all links starting with old_domain
    # We'll capture the path after old_domain and replace accordingly
    pattern = re.compile(r'<a\s+([^>]*?)href=["\']' + re.escape(old_domain) + r'([^"\']*)["\']([^>]*)>', re.IGNORECASE)

    def repl(match):
        before_href = match.group(1)
        path = match.group(2)
        after_href = match.group(3)
        # build new link with onclick and href="#"
        return f'<a {before_href}href="#" onclick="window.location = linkhost + \'{path}\'"{after_href}>'

    new_content = pattern.sub(repl, content)

    if new_content != content:
        with open(filepath, 'w', encoding='utf-8') as f:
            f.write(new_content)
        print(f"Updated links and added script in {filepath}")


def process_folder(folder):
    for root, dirs, files in os.walk(folder):
        for file in files:
            if file.endswith('.html'):
                filepath = os.path.join(root, file)
                add_script_and_replace_links(filepath)


if __name__ == "__main__":
    process_folder(root_folder)
    print("All done!")
```
